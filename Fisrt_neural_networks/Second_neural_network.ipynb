```python
fashion_mnist = tf.keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
```

```python
#   here  clearly  shows  that the 0 output means  that the  item is a ankle boot.  which label is 9 in the  fashion mnist dataset.

#  now  lets  see some  more examples of the  data in form of  images and  their labels

fig ,ax = plt.subplots(nrows=3, ncols=3, figsize=(10,10))
ax = ax.flatten()
for i in range(9):
    img = x_train[i]
    ax[i].imshow(img, cmap=plt.cm.binary)
    ax[i].set_title(f"Label: {y_train[i]}")

plt.tight_layout()
plt.show()
```
