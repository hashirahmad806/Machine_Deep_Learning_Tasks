````python
import os
from tensorflow.keras.utils import get_file
from tensorflow.keras.applications import ResNet50

# Create the models folder
os.makedirs("models", exist_ok=True)

# Download directly into the models folder
weights_path = get_file(
    fname="resnet50_weights_tf_dim_ordering_tf_kernels.h5",
    origin="https://storage.googleapis.com/tensorflow/keras-applications/resnet/resnet50_weights_tf_dim_ordering_tf_kernels.h5",
    cache_dir=os.getcwd(),
    cache_subdir="models"
)

print("Weights saved to:", weights_path)

# Load the pretrained model
model = ResNet50(weights=weights_path)

print("ResNet50 loaded successfully!")
````

````python
image_path = './dog1.webp'  # Replace with your image path`
loaded_image = keras_image_utils.load_img(image_path, target_size=(224, 224))
x = keras_image_utils.img_to_array(loaded_image)
# mean to add a new dimension to the array, making it suitable for model input
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)
````

````python
preds = model.predict(x)
# decode the results into a list of tuples (class, description, probability)
decoded_predictions = decode_predictions(preds, top=3)
print( "Prediction :: ", decoded_predictions)
````
